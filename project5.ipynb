```python
import spacy
import operator
from collections import Counter

# Project 5
# Text: Pride and Prejudice by Jane Austen
# Source: Project Gutenberg
# I chose this text because I wanted to analyze a classic novel and see what words appeared the most.

import requests

url = "https://www.gutenberg.org/files/1342/1342-0.txt"
text = requests.get(url).text

eng_processor = spacy.load("en_core_web_sm")

def word_tokenization_normalization(text):
    doc = eng_processor(text)

    cleaned_words = []

    for token in doc:
        if not token.is_stop and not token.is_punct and not token.like_num:
            if token.text.strip() != "":
                cleaned_words.append(token.lemma_.lower())

    return cleaned_words

def word_count(word_list):
    return Counter(word_list)

def print_top_15_frequent_words(word_counts):
    top_words = word_counts.most_common(15)

    print("Top 15 Most Frequent Words:\n")

    for word, count in top_words:
        print(word, ":", count)

cleaned_text = word_tokenization_normalization(text)

word_counts = word_count(cleaned_text)

print_top_15_frequent_words(word_counts)

# Extended analysis - most common verbs

def most_common_verbs(text):
    """
    Finds the most common verbs in the text.
    """

    doc = eng_processor(text)

    verbs = []

    for token in doc:
        if token.pos_ == "VERB" and not token.is_stop:
            verbs.append(token.lemma_.lower())

    verb_counts = Counter(verbs)

    return verb_counts.most_common(10)

print("\nTop 10 Most Common Verbs:\n")

for verb, count in most_common_verbs(text):
    print(verb, ":", count)
```
