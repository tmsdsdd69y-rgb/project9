```python
# Plot 1: Compare petal length by species using a box plot
plt.figure(figsize=(8, 5))
clean_df.boxplot(column="petal length (cm)", by="species")
plt.title("Petal Length by Iris Species")
plt.suptitle("")
plt.xlabel("Species")
plt.ylabel("Petal Length (cm)")
plt.show()
```

```python
# Plot 2: Compare petal length and petal width using a scatter plot
plt.figure(figsize=(8, 5))

for species in clean_df["species"].unique():
    species_data = clean_df[clean_df["species"] == species]
    plt.scatter(
        species_data["petal length (cm)"],
        species_data["petal width (cm)"],
        label=species
    )

plt.title("Petal Length vs Petal Width by Species")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.legend(title="Species")
plt.show()
```
